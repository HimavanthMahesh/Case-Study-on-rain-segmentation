### Dataset

```python
from pathlib import Path
import pandas as pd
import cv2
import numpy as np
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

DATASETS = {
    "clean": "cityscapes_subset_50/aachen",
    "rain_light_1": "rain-rendering/data/output/diff_l_25_v1/leftImg8bit/train/aachen/rain/25mm/rainy_image",
    "rain_light_2": "rain-rendering/data/output/diff_l_25_v2/leftImg8bit/train/aachen/rain/25mm/rainy_image",
    "rain_light_3": "rain-rendering/data/output/diff_l_25_v3/leftImg8bit/train/aachen/rain/25mm/rainy_image",
    "rain_medium_1": "rain-rendering/data/output/diff_m_75_v1/leftImg8bit/train/aachen/rain/75mm/rainy_image",
    "rain_medium_2": "rain-rendering/data/output/diff_m_75_v2/leftImg8bit/train/aachen/rain/75mm/rainy_image",
    "rain_medium_3": "rain-rendering/data/output/diff_m_75_v3/leftImg8bit/train/aachen/rain/75mm/rainy_image",
    "rain_heavy_1": "rain-rendering/data/output/diff_h_140_v1/leftImg8bit/train/aachen/rain/140mm/rainy_image",
    "rain_heavy_2": "rain-rendering/data/output/diff_h_140_v2/leftImg8bit/train/aachen/rain/140mm/rainy_image",
    "rain_heavy_3": "rain-rendering/data/output/diff_h_140_v3/leftImg8bit/train/aachen/rain/140mm/rainy_image",
}

OUTPUTS = {
    "udr_mixer": "derain_outputs/udr_mixer",
    "uformer": "derain_outputs/uformer",
}

def list_images(folder):
    exts = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
    return sorted([p for p in Path(folder).iterdir() if p.suffix.lower() in exts])

def read_rgb(path):
    img = cv2.imread(str(path), cv2.IMREAD_COLOR)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img

import cv2
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

def compute_metrics(clean_img, test_img):
    if test_img.shape[:2] != clean_img.shape[:2]:
        test_img = cv2.resize(
            test_img,
            (clean_img.shape[1], clean_img.shape[0]),
            interpolation=cv2.INTER_LINEAR
        )

    psnr = peak_signal_noise_ratio(clean_img, test_img, data_range=255)
    ssim = structural_similarity(clean_img, test_img, channel_axis=2, data_range=255)
    return psnr, ssim

clean_map = {p.name: str(p) for p in list_images(DATASETS["clean"])}

rows = []
for ds_name, ds_path in DATASETS.items():
    if ds_name == "clean":
        continue
    severity = ds_name.split("_")[1]
    version = ds_name.split("_")[-1]

    for rainy_path in list_images(ds_path):
        if rainy_path.name not in clean_map:
            continue
        rows.append({
            "dataset": ds_name,
            "severity": severity,
            "version": version,
            "image_name": rainy_path.name,
            "clean_path": clean_map[rainy_path.name],
            "rainy_path": str(rainy_path),
            "udr_mixer_path": str(Path(OUTPUTS["udr_mixer"]) / ds_name / rainy_path.name),
            "uformer_path": str(Path(OUTPUTS["uformer"]) / ds_name / rainy_path.name),
        })

pairs_df = pd.DataFrame(rows)
pairs_df.head()
```

```python
results = []

for _, row in pairs_df.iterrows():
    clean_img = read_rgb(row["clean_path"])
    rainy_img = read_rgb(row["rainy_path"])

    rainy_psnr, rainy_ssim = compute_metrics(clean_img, rainy_img)

    out = {
        "dataset": row["dataset"],
        "severity": row["severity"],
        "version": row["version"],
        "image_name": row["image_name"],
        "rainy_psnr": rainy_psnr,
        "rainy_ssim": rainy_ssim,
    }

    for model_name in ["udr_mixer", "uformer"]:
        p = Path(row[f"{model_name}_path"])
        if p.exists():
            img = read_rgb(p)
            psnr, ssim = compute_metrics(clean_img, img)
            out[f"{model_name}_psnr"] = psnr
            out[f"{model_name}_ssim"] = ssim

    results.append(out)

metrics_df = pd.DataFrame(results)
metrics_df.head()
```

```python
from pathlib import Path
import pandas as pd
import cv2
import numpy as np
from skimage.metrics import peak_signal_noise_ratio, structural_similarity

ROOT = Path("/path/to/project")

DATASETS = {
    "clean": ROOT / "cityscapes_subset_50/aachen",

    "rain_light_1": ROOT / "rain-rendering/data/output/diff_l_25_v1/leftImg8bit/train/aachen/rain/25mm/rainy_image",
    "rain_light_2": ROOT / "rain-rendering/data/output/diff_l_25_v2/leftImg8bit/train/aachen/rain/25mm/rainy_image",
    "rain_light_3": ROOT / "rain-rendering/data/output/diff_l_25_v3/leftImg8bit/train/aachen/rain/25mm/rainy_image",

    "rain_medium_1": ROOT / "rain-rendering/data/output/diff_m_75_v1/leftImg8bit/train/aachen/rain/75mm/rainy_image",
    "rain_medium_2": ROOT / "rain-rendering/data/output/diff_m_75_v2/leftImg8bit/train/aachen/rain/75mm/rainy_image",
    "rain_medium_3": ROOT / "rain-rendering/data/output/diff_m_75_v3/leftImg8bit/train/aachen/rain/75mm/rainy_image",

    "rain_heavy_1": ROOT / "rain-rendering/data/output/diff_h_140_v1/leftImg8bit/train/aachen/rain/140mm/rainy_image",
    "rain_heavy_2": ROOT / "rain-rendering/data/output/diff_h_140_v2/leftImg8bit/train/aachen/rain/140mm/rainy_image",
    "rain_heavy_3": ROOT / "rain-rendering/data/output/diff_h_140_v3/leftImg8bit/train/aachen/rain/140mm/rainy_image",
}

OUTPUTS = {
    "drsformer": {
        "base": ROOT / "derain_outputs/drsformer",
        "subdir": "Deraining",
    },
    "idt": {
        "base": ROOT / "derain_outputs/idt",
        "subdir": "",
    },
    "nerdrain": {
        "base": ROOT / "derain_outputs/nerdrain",
        "subdir": "",
    },
}


def list_images(folder):
    folder = Path(folder)
    exts = {".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"}
    if not folder.exists():
        print(f"WARNING: folder does not exist: {folder}")
        return []
    return sorted([p for p in folder.iterdir() if p.suffix.lower() in exts])


def read_rgb(path):
    img = cv2.imread(str(path), cv2.IMREAD_COLOR)
    if img is None:
        raise FileNotFoundError(f"Could not read image: {path}")
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img


def compute_metrics(clean_img, test_img):
    if test_img.shape[:2] != clean_img.shape[:2]:
        test_img = cv2.resize(
            test_img,
            (clean_img.shape[1], clean_img.shape[0]),
            interpolation=cv2.INTER_LINEAR,
        )
    psnr = peak_signal_noise_ratio(clean_img, test_img, data_range=255)
    ssim = structural_similarity(clean_img, test_img, channel_axis=2, data_range=255)
    return psnr, ssim


def output_path_for(model_name, ds_name, image_name):
    info = OUTPUTS[model_name]
    base = Path(info["base"])
    subdir = info["subdir"]
    if subdir:
        return base / ds_name / subdir / image_name
    else:
        return base / ds_name / image_name
clean_map = {p.name: str(p) for p in list_images(DATASETS["clean"])}
rows = []
for ds_name, ds_path in DATASETS.items():
    if ds_name == "clean":
        continue
    parts = ds_name.split("_")
    severity = parts[1]
    version = parts[-1]
    rainy_images = list_images(ds_path)
    for rainy_path in rainy_images:
        if rainy_path.name not in clean_map:
            continue
        row = {
            "dataset": ds_name,
            "severity": severity,
            "version": version,
            "image_name": rainy_path.name,
            "clean_path": clean_map[rainy_path.name],
            "rainy_path": str(rainy_path),
        }
        for model_name in OUTPUTS:
            row[f"{model_name}_path"] = str(
                output_path_for(model_name, ds_name, rainy_path.name)
            )

        rows.append(row)
pairs_df = pd.DataFrame(rows)

print("Total rows:", len(pairs_df))
pairs_df.head()
```

```python
import shutil
import subprocess
from pathlib import Path

UDR_REPO = Path("/path/to/project/UDR-S2Former_deraining")
UDR_INPUT = UDR_REPO / "image_demo/input_images"
UDR_OUTPUT = UDR_REPO / "image_demo/output_images"

for name, input_dir in RAIN_DATASETS_V23.items():
    final_output = Path(f"/path/to/project/udrs2former_outputs/{name}")
    print(f"\nRunning UDR-S2Former on {name}")
    shutil.rmtree(UDR_INPUT, ignore_errors=True)
    shutil.rmtree(UDR_OUTPUT, ignore_errors=True)
    UDR_INPUT.mkdir(parents=True, exist_ok=True)
    UDR_OUTPUT.mkdir(parents=True, exist_ok=True)
    shutil.rmtree(final_output, ignore_errors=True)
    final_output.mkdir(parents=True, exist_ok=True)
    for p in sorted(Path(input_dir).glob("*")):
        if p.is_file() and p.suffix.lower() in [".png", ".jpg", ".jpeg"]:
            shutil.copy2(p, UDR_INPUT / p.name)
    cmd = [
        "conda", "run", "--no-capture-output", "-n", "udrs2former_a100",
        "python", "demo.py",
        "-c", "config/demo.yaml",
    ]

    subprocess.run(cmd, cwd=str(UDR_REPO), check=True)
    kept = 0
    for p in sorted(UDR_OUTPUT.glob("*")):
        if p.is_file() and not p.name.startswith("cat_"):
            shutil.copy2(p, final_output / p.name)
            kept += 1

    print(name, "outputs:", kept, "restored images")
```
