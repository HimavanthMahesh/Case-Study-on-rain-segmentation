```python
LIGHT_DIR  = "mseg-semantic/temp_files/mseg-3m-480p_rainy_image_universal_ms"
MEDIUM_DIR = "mseg-semantic/temp_files/mseg-3m-480p_75mm_rainy_image_universal_ss"
IGNORE_LABEL = 255
N_CLASSES = 256
```

```python
def find_label_maps(root):
    gray = glob.glob(os.path.join(root, "**/gray/**/*.png"), recursive=True)
    if len(gray) > 0:
        return sorted(gray)
    return sorted(glob.glob(os.path.join(root, "**/*.png"), recursive=True))
light_paths  = find_label_maps(LIGHT_DIR)
medium_paths = find_label_maps(MEDIUM_DIR)
L = {os.path.basename(p): p for p in light_paths}
M = {os.path.basename(p): p for p in medium_paths}
common = sorted(set(L.keys()) & set(M.keys()))
print("Light label maps:", len(light_paths))
print("Medium label maps:", len(medium_paths))
print("Common files:", len(common))
print("Example common:", common[:5])
```

```python
def load_mask(path):
    arr = np.array(Image.open(path))
    if arr.ndim == 3:
        arr = arr[..., 0]
    return arr.astype(np.int64)

def confusion_matrix(gt, pred, n_classes, ignore_label=None):
    if ignore_label is not None:
        mask = gt != ignore_label
        gt = gt[mask]
        pred = pred[mask]
    valid = (gt >= 0) & (gt < n_classes) & (pred >= 0) & (pred < n_classes)
    gt = gt[valid]
    pred = pred[valid]
    idx = gt * n_classes + pred
    cm = np.bincount(idx, minlength=n_classes*n_classes).reshape(n_classes, n_classes)
    return cm

def miou_from_cm(cm):
    tp = np.diag(cm).astype(np.float64)
    fp = cm.sum(axis=0) - tp
    fn = cm.sum(axis=1) - tp
    denom = tp + fp + fn
    iou = np.divide(tp, denom, out=np.full_like(tp, np.nan), where=denom > 0)
    return float(np.nanmean(iou))

def rmse_ids(a, b, ignore_label=None):
    if ignore_label is not None:
        mask = a != ignore_label
        a = a[mask]
        b = b[mask]
    if a.size == 0:
        return np.nan
    d = (a.astype(np.float64) - b.astype(np.float64))
    return float(np.sqrt(np.mean(d*d)))
rows = []
for name in common:
    light = load_mask(L[name])
    med   = load_mask(M[name])
    if light.shape != med.shape:
        continue
    cm = confusion_matrix(med, light, N_CLASSES, ignore_label=IGNORE_LABEL)
    miou = miou_from_cm(cm)
    rmse = rmse_ids(med, light, ignore_label=IGNORE_LABEL)

    rows.append((name, miou, rmse))

print("Evaluated:", len(rows))
```

```python
import matplotlib.pyplot as plt
import numpy as np

miou_vals = np.array([r[1] for r in rows], dtype=float)
rmse_vals = np.array([r[2] for r in rows], dtype=float)

plt.figure()
plt.scatter(miou_vals, rmse_vals)
plt.xlabel("mIoU agreement (medium vs light)")
plt.ylabel("RMSE of label IDs (medium vs light)")
plt.title("Prediction shift with increased rain")
plt.grid(alpha=0.3)
plt.show()
```

```python
import os, glob
import numpy as np
from PIL import Image
import cv2

IGNORE_LABEL = 255
N_CLASSES = 195

def list_pngs(folder):
    return sorted(glob.glob(os.path.join(folder, "*.png")))

def map_by_name(paths):
    return {os.path.basename(p): p for p in paths}

def load_mask(path):
    arr = np.array(Image.open(path))
    if arr.ndim == 3:
        arr = arr[..., 0]
    return arr.astype(np.int64)

def confusion_matrix(gt, pred, n_classes, ignore_label=None):
    if ignore_label is not None:
        m = (gt != ignore_label)
        gt = gt[m]; pred = pred[m]
    valid = (gt >= 0) & (gt < n_classes) & (pred >= 0) & (pred < n_classes)
    gt = gt[valid]; pred = pred[valid]
    idx = gt * n_classes + pred
    cm = np.bincount(idx, minlength=n_classes*n_classes).reshape(n_classes, n_classes)
    return cm

def miou_from_cm(cm):
    tp = np.diag(cm).astype(np.float64)
    fp = cm.sum(axis=0) - tp
    fn = cm.sum(axis=1) - tp
    denom = tp + fp + fn
    iou = np.divide(tp, denom, out=np.full_like(tp, np.nan), where=(denom > 0))
    return float(np.nanmean(iou))

def rmse_ids(a, b, ignore_label=None):
    if ignore_label is not None:
        m = (a != ignore_label)
        a = a[m]; b = b[m]
    if a.size == 0:
        return np.nan
    d = (a.astype(np.float64) - b.astype(np.float64))
    return float(np.sqrt(np.mean(d*d)))
    
def eval_against_clean(clean_dir, other_dir):
    C = map_by_name(list_pngs(clean_dir))
    O = map_by_name(list_pngs(other_dir))
    common = sorted(set(C) & set(O))
    if len(common) == 0:
        raise RuntimeError(f"No matching filenames between\n{clean_dir}\nand\n{other_dir}")
    miou_list, rmse_list = [], []
    resized_ct = 0
    for name in common:
        gt = load_mask(C[name])
        pr = load_mask(O[name])
        if gt.shape != pr.shape:
            gt = cv2.resize(gt, (pr.shape[1], pr.shape[0]), interpolation=cv2.INTER_NEAREST)
            resized_ct += 1
        cm = confusion_matrix(gt, pr, N_CLASSES, IGNORE_LABEL)
        miou_list.append(miou_from_cm(cm))
        rmse_list.append(rmse_ids(gt, pr, IGNORE_LABEL))
    return {
        "n": len(miou_list),
        "resized": resized_ct,
        "miou_per_image": miou_list,
        "rmse_per_image": rmse_list,
        "mean_miou": float(np.nanmean(miou_list)),
        "mean_rmse": float(np.nanmean(rmse_list)),
    }
```

```python
import numpy as np
from PIL import Image
import os
import glob
from scipy.stats import mode
```

```python
def majority_vote_masks(folder_list, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    files = sorted(glob.glob(os.path.join(folder_list[0], "*.png")))
    for f in files:
        name = os.path.basename(f)
        masks = []
        for folder in folder_list:
            path = os.path.join(folder, name)
            arr = np.array(Image.open(path))
            if arr.ndim == 3:
                arr = arr[..., 0]
            masks.append(arr)
        stack = np.stack(masks, axis=0)
        voted = mode(stack, axis=0, keepdims=False)[0]
        Image.fromarray(voted.astype(np.uint8)).save(os.path.join(output_dir, name))
    print("Saved averaged masks to:", output_dir)
```

```python
import numpy as np
from PIL import Image
import os, glob

def majority_vote_3(a, b, c):
    return np.where((a == b) | (a == c), a, np.where(b == c, b, a))

def majority_vote_masks(runs, output_dir, max_images=50):
    os.makedirs(output_dir, exist_ok=True)
    files = sorted(glob.glob(os.path.join(runs[0], "*.png")))[:max_images]
    for p0 in files:
        name = os.path.basename(p0)
        p1 = os.path.join(runs[1], name)
        p2 = os.path.join(runs[2], name)
        if not (os.path.exists(p1) and os.path.exists(p2)):
            continue
        a = np.array(Image.open(p0))
        b = np.array(Image.open(p1))
        c = np.array(Image.open(p2))
        if a.ndim == 3: a = a[..., 0]
        if b.ndim == 3: b = b[..., 0]
        if c.ndim == 3: c = c[..., 0]
        voted = majority_vote_3(a, b, c).astype(np.uint8)
        Image.fromarray(voted).save(os.path.join(output_dir, name))

    print("Saved:", output_dir)
```

### Coloring and overlaying

```python
import os, glob
import numpy as np
from PIL import Image

def make_colormap(n=256, seed=123):
    rng = np.random.default_rng(seed)
    cmap = rng.integers(0, 255, size=(n, 3), dtype=np.uint8)
    cmap[0] = np.array([0, 0, 0], dtype=np.uint8)
    return cmap
COLORMAP = make_colormap(256, seed=123)
def load_gray_mask(mask_path):
    m = np.array(Image.open(mask_path))
    if m.ndim == 3:
        m = m[..., 0]
    return m.astype(np.int32)
def colorize_mask(mask, colormap=COLORMAP):
    mask = np.clip(mask, 0, colormap.shape[0]-1)
    return colormap[mask]
def overlay_on_image(img_path, mask_path, out_path, alpha=0.45):
    img = np.array(Image.open(img_path).convert("RGB"), dtype=np.uint8)
    mask = load_gray_mask(mask_path)
    rgb = colorize_mask(mask)
    if rgb.shape[:2] != img.shape[:2]:
        rgb_pil = Image.fromarray(rgb)
        rgb_pil = rgb_pil.resize((img.shape[1], img.shape[0]), resample=Image.NEAREST)
        rgb = np.array(rgb_pil, dtype=np.uint8)
    out = (img.astype(np.float32) * (1 - alpha) + rgb.astype(np.float32) * alpha).astype(np.uint8)
    Image.fromarray(out).save(out_path)
def make_overlays(dataset_key, img_dir, pred_gray_dir, out_root="overlays", max_images=50, alpha=0.45):
    os.makedirs(out_root, exist_ok=True)
    out_dir = os.path.join(out_root, dataset_key)
    os.makedirs(out_dir, exist_ok=True)
    img_paths = sorted(glob.glob(os.path.join(img_dir, "*.png")))[:max_images]
    pred_map = {os.path.basename(p): p for p in glob.glob(os.path.join(pred_gray_dir, "*.png"))}
    done = 0
    for ip in img_paths:
        name = os.path.basename(ip)
        if name not in pred_map:
            continue
        op = os.path.join(out_dir, name.replace(".png", "_overlay.png"))
        overlay_on_image(ip, pred_map[name], op, alpha=alpha)
        done += 1
    print(f"{dataset_key}: wrote {done} overlays to {out_dir}")
    return out_dir
```

```python
import os
import glob
import cv2
import numpy as np
import torch
import torch.nn.functional as F
from PIL import Image
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

if "model" not in globals():
    raise RuntimeError(
        "This softmax section expects an already-loaded MSeg model in a variable named `model`. "
        "Load the model first using your existing MSeg setup, then rerun this section."
    )

model = model.to(device)
model.eval()

def list_pngs(folder):
    return sorted(glob.glob(os.path.join(folder, "*.png")))

def common_png_names(folder_list):
    name_sets = []
    for folder in folder_list:
        name_sets.append({os.path.basename(p) for p in list_pngs(folder)})
    return sorted(set.intersection(*name_sets))

def load_rgb_image(path):
    img_bgr = cv2.imread(path, cv2.IMREAD_COLOR)
    if img_bgr is None:
        raise FileNotFoundError(f"Could not read image: {path}")
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    return img_rgb

def preprocess_image_for_mseg(img_rgb):
    x = img_rgb.astype(np.float32) / 255.0
    mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)
    std = np.array([0.229, 0.224, 0.225], dtype=np.float32)
    x = (x - mean) / std
    x = np.transpose(x, (2, 0, 1))
    x = torch.from_numpy(x).unsqueeze(0).to(device)
    return x

def extract_logits(output):
    if isinstance(output, torch.Tensor):
        logits = output
    elif isinstance(output, (list, tuple)):
        logits = output[0]
    elif isinstance(output, dict):
        for key in ["logits", "pred", "prediction", "out", "output"]:
            if key in output:
                logits = output[key]
                break
        else:
            raise ValueError(f"Could not find logits in dict output with keys: {list(output.keys())}")
    else:
        raise TypeError(f"Unsupported model output type: {type(output)}")

    if logits.ndim != 4:
        raise ValueError(f"Expected logits with shape [B,C,H,W], got {tuple(logits.shape)}")
    return logits

@torch.no_grad()
def predict_probs_for_image(img_path):
    img_rgb = load_rgb_image(img_path)
    orig_h, orig_w = img_rgb.shape[:2]
    x = preprocess_image_for_mseg(img_rgb)   # [1,3,H,W]
    _, _, h, w = x.shape
    new_h = ((h - 1 + 7) // 8) * 8 + 1
    new_w = ((w - 1 + 7) // 8) * 8 + 1
    pad_h = new_h - h
    pad_w = new_w - w
    if pad_h > 0 or pad_w > 0:
        x = F.pad(x, (0, pad_w, 0, pad_h), mode="replicate")
    output = model(x)
    logits = extract_logits(output)
    logits = F.interpolate(
        logits,
        size=(orig_h, orig_w),
        mode="bilinear",
        align_corners=False
    )
    probs = F.softmax(logits, dim=1)[0]   # [C,H,W]
    return probs

@torch.no_grad()
def softmax_average_triplet(img_paths):
    probs_list = [predict_probs_for_image(p) for p in img_paths]
    avg_probs = torch.stack(probs_list, dim=0).mean(dim=0)
    pred_mask = torch.argmax(avg_probs, dim=0)
    conf_map = torch.max(avg_probs, dim=0).values
    return avg_probs, pred_mask, conf_map

def save_mask_uint8(mask_tensor, out_path):
    mask = mask_tensor.detach().cpu().numpy().astype(np.uint8)
    Image.fromarray(mask).save(out_path)

def save_confidence_uint8(conf_tensor, out_path):
    conf = conf_tensor.detach().cpu().numpy()
    conf = np.clip(conf, 0.0, 1.0)
    conf = (conf * 255.0).astype(np.uint8)
    Image.fromarray(conf).save(out_path)
```

```python
def run_softmax_averaging_for_severity(severity_key, image_dirs, out_root=SOFTMAX_AVG_ROOT):
    common = common_png_names(image_dirs)
    if len(common) == 0:
        raise RuntimeError(f"No common filenames found for {severity_key}")

    gray_dir = os.path.join(out_root, f"{severity_key}_softmax_avg", "360", "gray")
    conf_dir = os.path.join(out_root, f"{severity_key}_softmax_avg", "360", "confidence")
    prob_dir = os.path.join(out_root, f"{severity_key}_softmax_avg", "360", "prob_npy")

    os.makedirs(gray_dir, exist_ok=True)
    os.makedirs(conf_dir, exist_ok=True)
    os.makedirs(prob_dir, exist_ok=True)

    for name in tqdm(common, desc=f"{severity_key} softmax avg"):
        img_paths = [os.path.join(d, name) for d in image_dirs]
        avg_probs, pred_mask, conf_map = softmax_average_triplet(img_paths)

        save_mask_uint8(pred_mask, os.path.join(gray_dir, name))
        save_confidence_uint8(conf_map, os.path.join(conf_dir, name))
        np.save(os.path.join(prob_dir, name.replace(".png", ".npy")), avg_probs.detach().cpu().numpy())

    return gray_dir, conf_dir, prob_dir

SOFTMAX_OUTPUTS = {}
for sev, dirs in SOFTMAX_IMAGE_RUNS.items():
    gray_dir, conf_dir, prob_dir = run_softmax_averaging_for_severity(sev, dirs)
    SOFTMAX_OUTPUTS[sev] = {
        "gray_dir": gray_dir,
        "conf_dir": conf_dir,
        "prob_dir": prob_dir,
    }
    print(sev, "-", gray_dir)
```

```python
import os
import numpy as np
from tqdm.auto import tqdm
import torch

@torch.no_grad()
def softmax_median_triplet(img_paths):
    probs_list = [predict_probs_for_image(p) for p in img_paths]   # each [C,H,W]
    stacked = torch.stack(probs_list, dim=0)                       # [3,C,H,W]
    med_probs = torch.median(stacked, dim=0).values                # [C,H,W]
    pred_mask = torch.argmax(med_probs, dim=0)                     # [H,W]
    conf_map = torch.max(med_probs, dim=0).values                  # [H,W]
    return med_probs, pred_mask, conf_map

def run_softmax_median_for_severity(severity_key, image_dirs, out_root="mseg-semantic/temp_files/rain_softmax_median_masks"):
    common = common_png_names(image_dirs)
    if len(common) == 0:
        raise RuntimeError(f"No common filenames found for {severity_key}")

    gray_dir = os.path.join(out_root, f"{severity_key}_softmax_median", "360", "gray")
    conf_dir = os.path.join(out_root, f"{severity_key}_softmax_median", "360", "confidence")
    prob_dir = os.path.join(out_root, f"{severity_key}_softmax_median", "360", "prob_npy")

    os.makedirs(gray_dir, exist_ok=True)
    os.makedirs(conf_dir, exist_ok=True)
    os.makedirs(prob_dir, exist_ok=True)

    for name in tqdm(common, desc=f"{severity_key} softmax median"):
        img_paths = [os.path.join(d, name) for d in image_dirs]
        med_probs, pred_mask, conf_map = softmax_median_triplet(img_paths)

        save_mask_uint8(pred_mask, os.path.join(gray_dir, name))
        save_confidence_uint8(conf_map, os.path.join(conf_dir, name))
        np.save(
            os.path.join(prob_dir, name.replace(".png", ".npy")),
            med_probs.detach().cpu().numpy()
        )

    return gray_dir, conf_dir, prob_dir
```

```python
import os
import glob
import cv2
import numpy as np
from PIL import Image

def list_pngs_filtered(folder, exclude_prefixes=("cat_",)):
    paths = sorted(glob.glob(os.path.join(folder, "*.png")))
    return [
        p for p in paths
        if not os.path.basename(p).startswith(exclude_prefixes)
    ]

def map_by_name_filtered(folder, exclude_prefixes=("cat_",)):
    paths = list_pngs_filtered(folder, exclude_prefixes=exclude_prefixes)
    return {os.path.basename(p): p for p in paths}

def eval_against_clean_filtered(clean_dir, other_dir, exclude_prefixes=("cat_",)):
    C = map_by_name_filtered(clean_dir, exclude_prefixes=exclude_prefixes)
    O = map_by_name_filtered(other_dir, exclude_prefixes=exclude_prefixes)

    common = sorted(set(C) & set(O))
    if len(common) == 0:
        raise RuntimeError(f"No matching filenames between\n{clean_dir}\nand\n{other_dir}")

    miou_list, rmse_list = [], []
    resized_ct = 0

    for name in common:
        gt = load_mask(C[name])
        pr = load_mask(O[name])

        if gt.shape != pr.shape:
            gt = cv2.resize(gt, (pr.shape[1], pr.shape[0]), interpolation=cv2.INTER_NEAREST)
            resized_ct += 1

        cm = confusion_matrix(gt, pr, N_CLASSES, IGNORE_LABEL)
        miou_list.append(miou_from_cm(cm))
        rmse_list.append(rmse_ids(gt, pr, IGNORE_LABEL))

    return {
        "n": len(miou_list),
        "resized": resized_ct,
        "miou_per_image": miou_list,
        "rmse_per_image": rmse_list,
        "mean_miou": float(np.nanmean(miou_list)),
        "mean_rmse": float(np.nanmean(rmse_list)),
    }
```

```python
severity_results = {}

severity_to_keys = {
    "light": {
        "Regular": PRED_DIRS["rain_light"],
        "Softmax": PRED_DIRS["rain_light_softmax_avg"],
        "Restormer": PRED_DIRS_DERAIN["restormer_light_v1"],
        "UDRS2Former": PRED_DIRS_DERAIN["udrs2former_light_v1"],
    },
    "medium": {
        "Regular": PRED_DIRS["rain_medium"],
        "Softmax": PRED_DIRS["rain_medium_softmax_avg"],
        "Restormer": PRED_DIRS_DERAIN["restormer_medium_v1"],
        "UDRS2Former": PRED_DIRS_DERAIN["udrs2former_medium_v1"],
    },
    "heavy": {
        "Regular": PRED_DIRS["rain_heavy"],
        "Softmax": PRED_DIRS["rain_heavy_softmax_avg"],
        "Restormer": PRED_DIRS_DERAIN["restormer_heavy_v1"],
        "UDRS2Former": PRED_DIRS_DERAIN["udrs2former_heavy_v1"],
    },
}

clean_dir = PRED_DIRS["clean"]

for severity, method_dirs in severity_to_keys.items():
    severity_results[severity] = {}

    print(f"\n{severity.upper()}")
    for label, pred_dir in method_dirs.items():
        if label == "UDRS2Former":
            r = eval_against_clean_filtered(
                clean_dir,
                pred_dir,
                exclude_prefixes=("cat_",)
            )
        else:
            r = eval_against_clean(clean_dir, pred_dir)

        severity_results[severity][label] = r

        print(
            label,
            "N=", r["n"],
            "resized=", r["resized"],
            "mean mIoU=", r["mean_miou"],
            "mean RMSE=", r["mean_rmse"],
        )
```
