```python
import cv2
import albumentations as A
import os
```

```python
transform = A.Compose([
    A.RandomRain(
        slant_lower=-20,
        slant_upper=20,
        drop_length=25,
        drop_width=1,
        drop_color=(200, 200, 200),
        blur_value=5,
        brightness_coefficient=0.7,
        rain_type="heavy"
    )
])

input_dir = "clean_images"
output_dir = "rain_images"
os.makedirs('/path/to/project', exist_ok=True)
```

```python
image_list = sorted(os.listdir(rain-rendering/data/source/cityscapes/leftImg8bit/train/aachen))[:50]
for img_name in image_list :
    img = cv2.imread(os.path.join(input_dir, img_name))
    augmented = transform(image=img)
    cv2.imwrite(os.path.join(output_dir, img_name), augmented["image"])
```
